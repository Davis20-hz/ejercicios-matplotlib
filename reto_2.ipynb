```python
def secuencia_cortes(n):
    resultados = []
    f = 1  # f(0)
    resultados.append(f)

    for i in range(1, n + 1):
        f = f + i #toma el valor anterior de f
#súmar el número del corte actual i
#guarda ese nuevo resultado como f(i)
        resultados.append(f)

    return resultados

# Elegir un valor de n=15, nos ayuda a ver la progresion exponencial sin llegar a un numero de cortes, que de ser mayor, seria muy dificil de imiginar o que simplemente no ocuurriria normalmente
n = 15
secuencia=secuencia_cortes(n)

print(f"Secuencia del cortador de pizza perezoso para n={n} cortes:")
for i, valor in enumerate(secuencia):
    print(f"f({i}) = {valor}")
```

```python
import math

def serie_madhava(n):
    pi_aprox = 0
    Aproximaciones = []
    for k in range(n):
        termino= ((-1)**k) / ((2*k + 1) * (3**k))
        pi_aprox += termino
        Aproximaciones.append(math.sqrt(12) * pi_aprox)
    return Aproximaciones

n_terms = 20
results = serie_madhava(n_terms)

print(f"Aproximaciones de pi usando la serie de Madhava (primeros {n_terms} términos):")
for i, aprox in enumerate(results):
    print(f"Termino {i+1}: {aprox}")
```
